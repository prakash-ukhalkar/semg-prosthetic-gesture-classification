# Section 8: Dataset Acquisition

## NinaPro Database (Primary Dataset)
The primary dataset selected for this project is the **NinaPro (Non-Invasive Collaborative Robotic Hand Prosthesis) Database**. 
NinaPro is an internationally recognized benchmark database for surface electromyography (sEMG) hand and wrist gesture classification.

### Why NinaPro is Selected:
- **Scientific Consensus**: Widely cited in literature (over 500 publications), enabling direct comparison of model accuracies.
- **High Channel Density**: Typically recorded using 8 to 12 active sEMG channels (e.g., Otto Bock MyoBock or Delsys Trigno systems).
- **Diverse Gestures**: Contains kinematic and muscular activations for 40+ distinct hand/wrist movements (including grasping, flexing, and extension).

### Official Access & Download:
- **Official Website**: http://ninapro.hevs.ch/
- **Licensing**: Access requires registering on the site and signing a research agreement. Therefore, automatic download cannot be implemented.
- **Target Subset**: **Database 2 (DB2)** is recommended for high-resolution studies (12 channels, 2000 Hz sampling rate, 40 subjects, 49 gestures).
- **Manual Step**: Please download the dataset MAT files for targeted subjects (e.g., `S1_E1_A1.mat`) and place them in the `data/raw/` directory.

### Expected Raw File Layout:
```text
data/raw/
├── DB2/
│   ├── S1_E1_A1.mat
│   ├── S1_E2_A1.mat
│   └── ...
└── .gitkeep
```

## Section 9: Dataset Verification

We scan the `data/raw/` folder for downloaded file sets to verify file counts, names, format extensions, and sizes.

````python
# Code Section - Dataset scan & optional acquisition
from src.utils import verify_raw_dataset, download_and_extract_subject

raw_dataset_stats = verify_raw_dataset(RAW_DATA_DIR)

print(f"Auditing raw data folder: {RAW_DATA_DIR}")
if raw_dataset_stats['file_count'] == 0:
    logger.info("Raw dataset empty. Attempting public download of Subject 1 from NinaPro DB2...")
    print("⚠ Raw dataset missing. Attempting public download of Subject 1 from NinaPro DB2...")
    success = download_and_extract_subject(subject_id=1, target_dir=RAW_DATA_DIR)
    if success:
        print("✔ Subject 1 successfully downloaded and extracted.")
        raw_dataset_stats = verify_raw_dataset(RAW_DATA_DIR)
    else:
        print("❌ Automated download failed. Please manually download from http://ninapro.hevs.ch/ and extract to data/raw/")
else:
    print("✔ Raw dataset files found.")

if raw_dataset_stats['exists'] and raw_dataset_stats['file_count'] > 0:
    logger.info(f"Raw dataset verification: {raw_dataset_stats['file_count']} files verified.")
    print(f"✔ Raw files detected: {raw_dataset_stats['file_count']} files")
    print(f"✔ Total Size: {raw_dataset_stats['total_size_mb']} MB")
    print(f"✔ File Formats: {raw_dataset_stats['file_formats']}")
````

## Section 10: Dataset Metadata

We export verified dataset statistics and file records as a JSON document to `outputs/reports/dataset_metadata.json` for integration into the research workflow.

````python
# Code Section - Metadata Export
import json

metadata = {
    "dataset_name": "NinaPro Database",
    "version": "DB2",
    "download_date": time.strftime('%Y-%m-%d'),
    "raw_directory": str(RAW_DATA_DIR),
    "file_count": raw_dataset_stats['file_count'],
    "total_size_mb": raw_dataset_stats['total_size_mb'],
    "file_formats": raw_dataset_stats['file_formats']
}

metadata_file_path = REPORTS_DIR / "dataset_metadata.json"
with open(metadata_file_path, "w", encoding="utf-8") as f:
    json.dump(metadata, f, indent=4)

logger.info(f"Dataset metadata written to: {metadata_file_path}")
print(f"✔ Metadata successfully written to: {metadata_file_path}")
````
